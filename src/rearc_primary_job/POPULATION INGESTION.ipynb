```python
import os
from datetime import datetime, timezone

import requests
```

```python
population_url = (
    "https://honolulu-api.datausa.io/tesseract/data.jsonrecords"
    "?cube=acs_yg_total_population_1&drilldowns=Year%2CNation&locale=en&measures=Population"
)
population_dir = "/Volumes/rearc/raw/volume/datausa/population"

population_response = requests.get(population_url, timeout=60)
population_response.raise_for_status()
population_response.json()  # fail here if the API returned something other than JSON

# The API has no last-modified date, so each run is saved as a new file named with the
# fetch time (UTC), e.g. .../population/population_20261005_143000.json.
os.makedirs(population_dir, exist_ok=True)
population_path = os.path.join(
    population_dir, f"population_{datetime.now(timezone.utc):%Y%m%d_%H%M%S}.json"
)
with open(population_path, "wb") as f:
    f.write(population_response.content)
print(f"Population data written to {population_path}")
```
